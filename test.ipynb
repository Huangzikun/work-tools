```python
line = "/Users/huangzikun/Desktop/桂林学院/算法设计与分析/22物联网-期末报告(附件)"

import os
print(os.path.exists(line))
```

```python
files = os.listdir(line)
files
```

```python
import zipfile

line_target = "/Users/huangzikun/Desktop/桂林学院/算法设计与分析/22物联网-期末报告"

# 循环遍历 files 列表
for file in files:
    file_path = os.path.join(line, file)
    # 检查文件是否为 zip 文件
    if file.lower().endswith('.zip'):
        try:
            # 去掉 .zip 后缀
            new_folder_name = os.path.splitext(file)[0]
            # 构建新的解压目录
            new_extract_path = os.path.join(line_target, new_folder_name)
            # 创建新的解压目录
            os.makedirs(new_extract_path, exist_ok=True)
            # 打开 zip 文件
            with zipfile.ZipFile(file_path, 'r') as zip_ref:
                # 解压文件到新的目录
                zip_ref.extractall(new_extract_path)
            print(f"成功解压 {file} 到 {new_extract_path}")
        except Exception as e:
            print(f"解压 {file} 时出错: {e}")
```
